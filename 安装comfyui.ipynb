```python
import subprocess
def sh(c):
    r = subprocess.run(c, shell=True, capture_output=True, text=True); return (r.stdout or r.stderr).strip()

# 1. 用 == 锁定现有版本
open("/tmp/constraints.txt", "w").write(
"""torch==2.12.0+git6bbd260
torchaudio==2.11.0+34c52a6
torchvision==0.27.1+df56172
triton==3.7.1+gitf0b55c07
""")

# 2. 更新 ComfyUI(已存在,不再 clone)
print(sh("cd /mnt/workspace/ComfyUI && git pull 2>&1 | tail -3"))

# 3. 安装依赖,跳过 torch 行并加约束
print(sh("cd /mnt/workspace/ComfyUI && grep -viE '^(torch|torchvision|torchaudio)([=<>~ ]|$)' requirements.txt > /tmp/req.txt "
         "&& pip install -r /tmp/req.txt -c /tmp/constraints.txt 2>&1 | tail -10"))

# 4. 验证 torch 系列没被改动且能 import
print(sh("pip freeze 2>/dev/null | grep -iE '^(torch|torchvision|torchaudio|triton)'"))
print(sh("python -c \"import torch, torchvision, torchaudio; print(torch.__version__, torch.version.hip, torch.cuda.is_available())\""))
```
