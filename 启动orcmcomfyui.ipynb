```python
import subprocess
logf = open("/mnt/workspace/comfy.log", "w")
comfy = subprocess.Popen(
    "cd /mnt/workspace/ComfyUI && python main.py --listen 0.0.0.0 --port 8188",
    shell=True, stdout=logf, stderr=subprocess.STDOUT)

import time; time.sleep(30)
print(open("/mnt/workspace/comfy.log").read()[-2500:])
```

```python
print(open("/mnt/workspace/comfy.log").read()[-1200:])
print("进程状态:", "运行中" if comfy.poll() is None else f"已退出,返回码 {comfy.poll()}")
```

```python
#G关闭
import subprocess, time

try:
    comfy.terminate()
    comfy.wait(timeout=15)
except subprocess.TimeoutExpired:
    comfy.kill()
    comfy.wait()
except NameError:
    pass   # 内核重启过,comfy 变量丢了,走下面的按端口杀
finally:
    try: logf.close()
    except Exception: pass

# 因为是 shell=True 启动的,python main.py 可能是子进程,再按命令行兜底清理
subprocess.run("pkill -f 'ComfyUI.*main.py|python main.py --listen 0.0.0.0 --port 8188'", shell=True)
time.sleep(3)

# 确认显存已释放
print(subprocess.run("rocm-smi --showmeminfo vram", shell=True, capture_output=True, text=True).stdout)
```
