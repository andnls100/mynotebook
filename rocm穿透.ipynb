```python
import os, re, time, subprocess, urllib.request

CF, CF_LOG = "/mnt/workspace/cloudflared/cloudflared", "/mnt/workspace/cloudflared.log"
if not os.path.exists(CF):
    urllib.request.urlretrieve(
        "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64", CF)
    os.chmod(CF, 0o755)

tun = subprocess.Popen([CF, "tunnel", "--url", "http://127.0.0.1:8188", "--no-autoupdate"],
                       stdout=open(CF_LOG, "w"), stderr=subprocess.STDOUT)
for _ in range(40):
    m = re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", open(CF_LOG).read())
    if m:
        print("外网地址:", m.group(0)); break
    time.sleep(1)
else:
    print("没拿到地址,日志:\n", open(CF_LOG).read()[-1500:])
```
